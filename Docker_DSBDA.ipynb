```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Ingest the Big Data
print("Loading server logs...")
df = pd.read_csv('/raw_logs.csv')

# 2. Data Cleaning: Convert JMeter's raw Unix timestamp into readable time
df['datetime'] = pd.to_datetime(df['timeStamp'], unit='ms')

# --- GRAPH 1: The Traffic Jam (Latency Analysis) ---
plt.figure(figsize=(12, 5))
# We plot the exact time of the request vs how many milliseconds it took to respond
plt.plot(df['datetime'], df['elapsed'], color='blue', alpha=0.7, linewidth=2)

plt.title('System Bottleneck: API Latency Over Time', fontsize=14, fontweight='bold')
plt.xlabel('Time of Request', fontsize=12)
plt.ylabel('Response Time (milliseconds)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

# Highlight the danger zone (Anything over 1000ms is a bad user experience)
plt.axhline(y=1000, color='red', linestyle='--', label='Critical Latency Threshold (1000ms)')
plt.legend()
plt.show()

print("\n" + "="*50 + "\n")

# --- GRAPH 2: The Casualties (Error Rate Analysis) ---
plt.figure(figsize=(8, 5))

# Convert to string to avoid numeric/string mixing errors
status_counts = df['responseCode'].astype(str).value_counts()

if len(status_counts) > 0:
    # Dynamically map green for 200, red for anything else
    bar_colors = ['#2ecc71' if code == '200' else '#e74c3c' for code in status_counts.index]

    ax = status_counts.plot(kind='bar', color=bar_colors)
    plt.title('High-Velocity Data Loss: HTTP Status Distribution', fontsize=14, fontweight='bold')
    plt.xlabel('HTTP Response Code', fontsize=12)
    plt.ylabel('Total Number of Requests', fontsize=12)
    plt.xticks(rotation=0)

    # Add the exact numbers on top of the bars safely
    for p in ax.patches:
        ax.annotate(str(int(p.get_height())), (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='bottom', fontweight='bold')
    plt.show()
else:
    print("⚠️ No response code data found in the CSV!")

print("\n" + "="*50 + "\n")
```

```python
# --- GRAPH 3: The Spread (Response Time Distribution) ---
plt.figure(figsize=(10, 6))

# A histogram groups the latencies into "buckets" (bins) to show frequency
plt.hist(df['elapsed'], bins=50, color='purple', alpha=0.7, edgecolor='black')

plt.title('Data Distribution: Frequency of Response Times', fontsize=14, fontweight='bold')
plt.xlabel('Response Time (milliseconds)', fontsize=12)
plt.ylabel('Number of Requests (Frequency)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
print("\n" + "="*50 + "\n")
```

```python
# --- GRAPH 4: Throughput Velocity (Requests Per Second) ---
plt.figure(figsize=(12, 5))

# We use Pandas to group the data by every single second ('S') and count the rows
# Note: df['datetime'] must be set as the index for resample to work
df_indexed = df.set_index('datetime')
throughput = df_indexed.resample('S').size()

plt.plot(throughput.index, throughput.values, color='teal', linewidth=2)

plt.title('System Throughput: Requests Handled Per Second', fontsize=14, fontweight='bold')
plt.xlabel('Time', fontsize=12)
plt.ylabel('Requests Per Second (RPS)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

# Fill the area under the curve for a more professional look
plt.fill_between(throughput.index, throughput.values, color='teal', alpha=0.2)

plt.show()
print("\n" + "="*50 + "\n")
```

```python
# --- GRAPH 5: System Fault Tolerance (Reliability Breakdown) ---
plt.figure(figsize=(8, 8))

# Standardize the text to lowercase ('true' vs 'True')
success_counts = df['success'].astype(str).str.lower().value_counts()

if len(success_counts) > 0:
    # Dynamically create labels and colors based on exactly what is in the data
    labels = ['Success (200 OK)' if val == 'true' else 'Failure (Error)' for val in success_counts.index]
    pie_colors = ['#2ecc71' if val == 'true' else '#e74c3c' for val in success_counts.index]

    # Dynamically set the 'explode' effect so it matches the number of slices
    explode_dynamic = [0.05] * len(success_counts)

    plt.pie(success_counts, labels=labels, autopct='%1.1f%%', startangle=140,
            colors=pie_colors, explode=explode_dynamic, shadow=True,
            textprops={'fontsize': 12, 'weight': 'bold'})

    plt.title('Overall System Reliability: Success vs. Failure Rate', fontsize=14, fontweight='bold')
    plt.show()
else:
    print("⚠️ No success/failure data found in the CSV!")
```
